### Hierarchical Multi-Head CNN Model for Food Classification (Fresh Produce vs Animal Protein)

```python
# ============================================================
# CNN MODEL 1 — HIERARCHICAL MULTI-HEAD 
# ============================================================
# ============================================================
# SECTION 1 — ENVIRONMENT SETUP & EXPERIMENT CONFIGURATION
# ============================================================
# WHAT THIS SECTION DOES:
# - Imports libraries
# - Sets device (GPU if available)
# - Defines config (dataset path, hyperparameters)
# - Fixes random seeds for reproducibility
# ============================================================

import os
import random
from dataclasses import dataclass
from typing import List, Dict, Tuple
from pathlib import Path
from collections import defaultdict
import math
import warnings

import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader

from torchvision import transforms
from PIL import Image

from tqdm import tqdm
import matplotlib.pyplot as plt

from sklearn.metrics import confusion_matrix, classification_report, accuracy_score

warnings.filterwarnings(
    "ignore",
    message="Palette images with Transparency expressed in bytes should be converted to RGBA images"
)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")


# ----------------------------
# Config: tunable settings
# ----------------------------
@dataclass
class Config:
    root_dir: str = r"C:\Users\Yioulika Antoniades\Desktop\UDA_Datasets\Dataset1_Classes"
    img_size: int = 160
    batch_size: int = 8
    num_workers: int = 0
    epochs: int = 35

    # AdamW settings
    lr: float = 1e-3
    weight_decay: float = 1e-4

    # Cosine scheduler settings
    min_lr: float = 1e-5

    out_dir: str = "outputs"
    num_example_preds: int = 12
    seed: int = 42
    normalize_labels: bool = True


cfg = Config()


def seed_everything(seed: int = 42):
    """
    Fix random seeds so you can reproduce results more reliably.
    """
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


seed_everything(cfg.seed)
os.makedirs(cfg.out_dir, exist_ok=True)


# ============================================================
# SECTION 2 — LABEL HIERARCHY & DATASET INTEGRITY CHECK
# ============================================================
# WHAT THIS SECTION DOES:
# - Defines hierarchy: Level1 -> Level2 -> Leaf
# - Builds label->index mappings
# - Prints dataset size and class distribution
# ============================================================

LEVEL1 = ["Fresh Produce", "Animal Protein"]
LEVEL2_FRESH = ["Fruits", "Vegetables"]
LEVEL2_ANIMAL = ["Meat & Poultry", "Seafood"]

FRUITS = ["Apples", "Avocados", "Bananas", "Grapes", "Mangos"]
VEGETABLES = ["Carrots", "Cucumbers", "Lettuce", "Potatoes", "Tomatoes"]
MEAT_POULTRY = ["Beef", "Chicken"]
SEAFOOD = ["Fish"]

IMAGE_EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".gif", ".tiff", ".webp"}


def _norm(s: str) -> str:
    """Normalize label strings to reduce folder naming issues (case/space)."""
    return " ".join(s.strip().lower().split())


def _maybe_norm(s: str, enabled: bool) -> str:
    return _norm(s) if enabled else s


LEAF_GLOBAL = FRUITS + VEGETABLES + MEAT_POULTRY + SEAFOOD

leaf_to_global = {_maybe_norm(name, cfg.normalize_labels): i for i, name in enumerate(LEAF_GLOBAL)}
global_to_leaf = {i: name for i, name in enumerate(LEAF_GLOBAL)}

level1_to_idx = {_maybe_norm(name, cfg.normalize_labels): i for i, name in enumerate(LEVEL1)}
level2_fresh_to_idx = {_maybe_norm(name, cfg.normalize_labels): i for i, name in enumerate(LEVEL2_FRESH)}
level2_animal_to_idx = {_maybe_norm(name, cfg.normalize_labels): i for i, name in enumerate(LEVEL2_ANIMAL)}

fruit_to_idx = {_maybe_norm(name, cfg.normalize_labels): i for i, name in enumerate(FRUITS)}
veg_to_idx = {_maybe_norm(name, cfg.normalize_labels): i for i, name in enumerate(VEGETABLES)}
meat_to_idx = {_maybe_norm(name, cfg.normalize_labels): i for i, name in enumerate(MEAT_POULTRY)}
seafood_to_idx = {_maybe_norm(name, cfg.normalize_labels): i for i, name in enumerate(SEAFOOD)}


def list_images_pathlib(folder: Path) -> List[Path]:
    """Recursively list image files."""
    if not folder.exists():
        return []
    return [p for p in folder.rglob("*") if p.is_file() and p.suffix.lower() in IMAGE_EXTS]


def human_bytes(num_bytes: int) -> str:
    units = ["B", "KB", "MB", "GB", "TB"]
    size = float(num_bytes)
    for u in units:
        if size < 1024 or u == units[-1]:
            return f"{size:.2f} {u}"
        size /= 1024


def folder_size_bytes(folder: Path) -> int:
    if not folder.exists():
        return 0
    total = 0
    for p in folder.rglob("*"):
        if p.is_file():
            try:
                total += p.stat().st_size
            except OSError:
                pass
    return total


def leaf_class_counts(split_dir: Path):
    images = list_images_pathlib(split_dir)
    counts = defaultdict(int)
    for img in images:
        counts[img.parent] += 1
    rel_counts = {str(folder.relative_to(split_dir)): n for folder, n in counts.items()}
    return rel_counts, len(images)


def inspect_dataset(dataset_root_str: str):
    dataset_root = Path(dataset_root_str)
    print("\nDataset successfully loaded from local directory.")
    print(f"Dataset root path: {dataset_root}")

    splits = {"train": dataset_root / "train", "val": dataset_root / "val", "test": dataset_root / "test"}

    missing = [name for name, p in splits.items() if not p.exists()]
    if missing:
        print("\nWARNING: Missing split folders:")
        for name in missing:
            print(f" - {name}: {splits[name]}")
        return

    total_size = folder_size_bytes(dataset_root)
    print(f"\nTotal dataset size: {human_bytes(total_size)}")

    grand_total_images = 0
    leaf_classes_union = set()

    for split_name, split_dir in splits.items():
        rel_counts, split_total = leaf_class_counts(split_dir)
        grand_total_images += split_total
        leaf_classes_union.update(rel_counts.keys())

        print(f"\n- {split_name.upper()} : {split_total} images")
        for rel_path in sorted(rel_counts.keys()):
            print(f"  - {rel_path} : {rel_counts[rel_path]} images")

    print("\nSummary:")
    print(f"  Total leaf classes (across splits): {len(leaf_classes_union)}")
    print(f"  Total images: {grand_total_images}")


inspect_dataset(cfg.root_dir)


# ============================================================
# SECTION 3 — DATASET LOADER & IMAGE PREPROCESSING PIPELINE
# ============================================================
# WHAT THIS SECTION DOES:
# - Defines Dataset class that parses folder structure into hierarchical labels
# - Computes mean/std from TRAIN split
# - Defines Train transforms (augmentation) and Val/Test transforms (deterministic)
# ============================================================

def list_images_for_split(split_dir: str) -> List[str]:
    return [str(x) for x in list_images_pathlib(Path(split_dir))]


class HierarchicalFoodDataset(Dataset):
    """
    Dataset structure:
      split/Level1/Level2/Leaf/image

    Returns:
      img tensor
      y1 (level 1)
      y2 (level 2)
      branch_id (0 fruit, 1 veg, 2 meat, 3 seafood)
      y_leaf_branch (leaf index within chosen branch)
      y_leaf_global (leaf index in LEAF_GLOBAL)
      meta dict (original folder names and file path for debugging/plots)
    """

    def __init__(self, split_dir: str, transform=None, normalize_labels: bool = True):
        self.split_dir = split_dir
        self.transform = transform
        self.normalize_labels = normalize_labels
        self.samples = list_images_for_split(split_dir)
        if len(self.samples) == 0:
            raise ValueError(f"No images found under: {split_dir}")

        self.parsed = []
        for p in self.samples:
            rel = os.path.relpath(p, split_dir)
            parts = rel.split(os.sep)
            if len(parts) < 4:
                raise ValueError(f"Bad path (expected Level1/Level2/Leaf/image): {p}")

            lvl1_raw, lvl2_raw, leaf_raw = parts[0], parts[1], parts[2]
            lvl1 = _maybe_norm(lvl1_raw, self.normalize_labels)
            lvl2 = _maybe_norm(lvl2_raw, self.normalize_labels)
            leaf = _maybe_norm(leaf_raw, self.normalize_labels)

            # Level 1 label
            if lvl1 not in level1_to_idx:
                raise ValueError(f"Unexpected Level1 '{lvl1_raw}' in: {p}")
            y1 = level1_to_idx[lvl1]

            # Level 2 label depends on which Level 1 branch we are in
            if y1 == 0:
                if lvl2 not in level2_fresh_to_idx:
                    raise ValueError(f"Unexpected Level2 '{lvl2_raw}' under Fresh Produce in: {p}")
                y2 = level2_fresh_to_idx[lvl2]
            else:
                if lvl2 not in level2_animal_to_idx:
                    raise ValueError(f"Unexpected Level2 '{lvl2_raw}' under Animal Protein in: {p}")
                y2 = level2_animal_to_idx[lvl2]

            # Decide which leaf head applies
            lvl1_fresh = _maybe_norm("Fresh Produce", self.normalize_labels)
            lvl1_animal = _maybe_norm("Animal Protein", self.normalize_labels)
            lvl2_fruits = _maybe_norm("Fruits", self.normalize_labels)
            lvl2_vegs = _maybe_norm("Vegetables", self.normalize_labels)
            lvl2_meat = _maybe_norm("Meat & Poultry", self.normalize_labels)
            lvl2_sea = _maybe_norm("Seafood", self.normalize_labels)

            if lvl1 == lvl1_fresh and lvl2 == lvl2_fruits:
                y_leaf_branch = fruit_to_idx[leaf]; branch_id = 0
            elif lvl1 == lvl1_fresh and lvl2 == lvl2_vegs:
                y_leaf_branch = veg_to_idx[leaf]; branch_id = 1
            elif lvl1 == lvl1_animal and lvl2 == lvl2_meat:
                y_leaf_branch = meat_to_idx[leaf]; branch_id = 2
            elif lvl1 == lvl1_animal and lvl2 == lvl2_sea:
                y_leaf_branch = seafood_to_idx[leaf]; branch_id = 3
            else:
                raise ValueError(f"Invalid hierarchy in: {p}")

            # Global leaf label
            if leaf not in leaf_to_global:
                raise ValueError(f"Leaf '{leaf_raw}' not in global leaf list. Check naming: {p}")
            y_leaf_global = leaf_to_global[leaf]

            meta = {"path": p, "lvl1": lvl1_raw, "lvl2": lvl2_raw, "leaf": leaf_raw}
            self.parsed.append((p, y1, y2, branch_id, y_leaf_branch, y_leaf_global, meta))

    def __len__(self):
        return len(self.parsed)

    def __getitem__(self, idx: int):
        p, y1, y2, branch_id, y_leaf_branch, y_leaf_global, meta = self.parsed[idx]
        img = Image.open(p).convert("RGB")
        if self.transform:
            img = self.transform(img)
        return (
            img,
            torch.tensor(y1, dtype=torch.long),
            torch.tensor(y2, dtype=torch.long),
            torch.tensor(branch_id, dtype=torch.long),
            torch.tensor(y_leaf_branch, dtype=torch.long),
            torch.tensor(y_leaf_global, dtype=torch.long),
            meta
        )


@torch.no_grad()
def compute_train_mean_std(train_dir: str, img_size: int, batch_size: int, num_workers: int):
    """
    Compute mean/std from the training split only (standard practice).
    Uses deterministic resizing/cropping so stats are stable.
    """
    base_tf = transforms.Compose([
        transforms.Resize(img_size + 16),
        transforms.CenterCrop(img_size),
        transforms.ToTensor(),
    ])
    ds = HierarchicalFoodDataset(train_dir, transform=base_tf, normalize_labels=cfg.normalize_labels)
    loader = DataLoader(ds, batch_size=batch_size, shuffle=False, num_workers=num_workers)

    channel_sum = torch.zeros(3)
    channel_sum_sq = torch.zeros(3)
    n_pixels = 0

    for x, *_ in tqdm(loader, desc="Computing mean/std", unit="batch"):
        b, _, h, w = x.shape
        n_pixels += b * h * w
        channel_sum += x.sum(dim=[0, 2, 3])
        channel_sum_sq += (x ** 2).sum(dim=[0, 2, 3])

    mean = channel_sum / n_pixels
    var = (channel_sum_sq / n_pixels) - (mean ** 2)
    std = torch.sqrt(var.clamp(min=1e-12))
    return mean.tolist(), std.tolist()


def build_transforms(mean: List[float], std: List[float]):
    """
    TRAIN PIPELINE:
      - RandomResizedCrop / flips / rotation / jitter
      - Normalize(mean,std)

    VAL/TEST PIPELINE:
      - Resize + CenterCrop (deterministic)
      - Normalize(mean,std)
    """
    train_tfms = transforms.Compose([
        transforms.RandomResizedCrop(cfg.img_size),
        transforms.RandomHorizontalFlip(p=0.5),
        transforms.RandomRotation(degrees=10),
        transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.05),
        transforms.ToTensor(),
        transforms.Normalize(mean=mean, std=std),
    ])
    test_tfms = transforms.Compose([
        transforms.Resize(cfg.img_size + 16),
        transforms.CenterCrop(cfg.img_size),
        transforms.ToTensor(),
        transforms.Normalize(mean=mean, std=std),
    ])
    return train_tfms, test_tfms


# ============================================================
# SECTION 4 — MODEL ARCHITECTURE & PREDICTION DECODING
# ============================================================
# WHAT THIS SECTION DOES:
# - Defines HierarchicalCNN with shared backbone + multiple heads
# - Defines prediction decoding function to get human-readable labels
# ============================================================

class HierarchicalCNN(nn.Module):
    """
    Shared CNN features + multi-head classifier:
      head1 -> Level 1 (2 classes)
      head2 -> Level 2 (2 classes)
      leaf heads -> one per branch (fruits/veg/meat/seafood)
    """
    def __init__(self):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 32, 3, padding=1), nn.BatchNorm2d(32), nn.ReLU(inplace=True),
            nn.MaxPool2d(2),
            nn.Conv2d(32, 64, 3, padding=1), nn.BatchNorm2d(64), nn.ReLU(inplace=True),
            nn.MaxPool2d(2),
            nn.Conv2d(64, 128, 3, padding=1), nn.BatchNorm2d(128), nn.ReLU(inplace=True),
            nn.MaxPool2d(2),
            nn.Conv2d(128, 256, 3, padding=1), nn.BatchNorm2d(256), nn.ReLU(inplace=True),
            nn.MaxPool2d(2),
        )
        self.pool = nn.AdaptiveAvgPool2d((1, 1))
        self.embed = nn.Sequential(
            nn.Flatten(),
            nn.Linear(256, 256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
        )
        self.head1 = nn.Linear(256, 2)
        self.head2 = nn.Linear(256, 2)
        self.fruit_head = nn.Linear(256, len(FRUITS))
        self.veg_head = nn.Linear(256, len(VEGETABLES))
        self.meat_head = nn.Linear(256, len(MEAT_POULTRY))
        self.seafood_head = nn.Linear(256, len(SEAFOOD))

    def forward(self, x):
        x = self.features(x)
        x = self.pool(x)
        h = self.embed(x)
        return (
            self.head1(h), self.head2(h),
            self.fruit_head(h), self.veg_head(h),
            self.meat_head(h), self.seafood_head(h)
        )


@torch.no_grad()
def predict_hierarchy_from_outputs(out1, out2, lf, lv, lm, ls):
    """
    Converts head logits into:
      - predicted Level1 index
      - predicted Level2 index
      - predicted global leaf index
      - predicted strings (Level1, Level2, Leaf)
    """
    pred1 = out1.argmax(dim=1)
    pred2 = out2.argmax(dim=1)

    pred_leaf_global = []
    pred_strings = []

    for i in range(out1.size(0)):
        p1, p2 = pred1[i].item(), pred2[i].item()

        if p1 == 0:
            lvl1_str = "Fresh Produce"
            if p2 == 0:
                lvl2_str = "Fruits"
                leaf_str = FRUITS[lf[i].argmax().item()]
            else:
                lvl2_str = "Vegetables"
                leaf_str = VEGETABLES[lv[i].argmax().item()]
        else:
            lvl1_str = "Animal Protein"
            if p2 == 0:
                lvl2_str = "Meat & Poultry"
                leaf_str = MEAT_POULTRY[lm[i].argmax().item()]
            else:
                lvl2_str = "Seafood"
                leaf_str = SEAFOOD[ls[i].argmax().item()]

        leaf_key = _maybe_norm(leaf_str, cfg.normalize_labels)
        pred_leaf_global.append(leaf_to_global[leaf_key])
        pred_strings.append((lvl1_str, lvl2_str, leaf_str))

    return pred1.cpu().numpy(), pred2.cpu().numpy(), np.array(pred_leaf_global), pred_strings


@torch.no_grad()
def predict_hierarchy(model, x):
    out1, out2, lf, lv, lm, ls = model(x)
    return predict_hierarchy_from_outputs(out1, out2, lf, lv, lm, ls)


# ============================================================
# SECTION 4.6 — TRAINING & EVALUATION LOOPS
# ============================================================

def train_one_epoch(model, loader, optimizer, criterion, epoch_idx, total_epochs):
    """
    TRAIN LOOP:
      - forward pass
      - compute losses for level1, level2, and correct leaf head
      - backprop + optimizer step
    """
    model.train()
    running_loss = 0.0
    pbar = tqdm(loader, desc=f"Epoch {epoch_idx}/{total_epochs} [train]", unit="batch")

    for x, y1, y2, branch_id, y_leaf_branch, _, _ in pbar:
        x = x.to(device); y1 = y1.to(device); y2 = y2.to(device)
        branch_id = branch_id.to(device); y_leaf_branch = y_leaf_branch.to(device)

        optimizer.zero_grad(set_to_none=True)
        out1, out2, lf, lv, lm, ls = model(x)

        L1 = criterion(out1, y1)
        L2 = criterion(out2, y2)

        # Leaf loss computed only for the appropriate branch head
        leaf_losses = []
        for i in range(x.size(0)):
            b = branch_id[i].item()
            head = [lf, lv, lm, ls][b]
            leaf_losses.append(criterion(head[i:i+1], y_leaf_branch[i:i+1]))

        L3 = torch.stack(leaf_losses).mean()
        loss = L1 + L2 + 2.0 * L3

        loss.backward()
        optimizer.step()

        running_loss += loss.item()
        pbar.set_postfix(loss=f"{loss.item():.4f}")

    return running_loss / max(1, len(loader))


@torch.no_grad()
def evaluate(model, loader, criterion):
    """
    EVALUATION LOOP:
      - no gradients
      - computes total loss + accuracies for level1/level2/leaf
      - returns arrays for confusion matrix + classification report
    """
    model.eval()
    running_loss = 0.0
    y1_true, y1_pred, y2_true, y2_pred, leaf_true, leaf_pred = [], [], [], [], [], []

    for x, y1, y2, branch_id, y_leaf_branch, y_leaf_global, _ in tqdm(
        loader, desc="Evaluating", unit="batch", leave=False
    ):
        x = x.to(device); y1 = y1.to(device); y2 = y2.to(device)
        branch_id = branch_id.to(device); y_leaf_branch = y_leaf_branch.to(device)

        out1, out2, lf, lv, lm, ls = model(x)

        L1 = criterion(out1, y1)
        L2 = criterion(out2, y2)

        leaf_losses = []
        for i in range(x.size(0)):
            b = branch_id[i].item()
            head = [lf, lv, lm, ls][b]
            leaf_losses.append(criterion(head[i:i+1], y_leaf_branch[i:i+1]))
        L3 = torch.stack(leaf_losses).mean()

        loss = L1 + L2 + 2.0 * L3
        running_loss += loss.item()

        p1 = out1.argmax(dim=1).cpu().numpy()
        p2 = out2.argmax(dim=1).cpu().numpy()
        _, _, p_leaf, _ = predict_hierarchy_from_outputs(out1, out2, lf, lv, lm, ls)

        y1_true.extend(y1.cpu().numpy().tolist()); y1_pred.extend(p1.tolist())
        y2_true.extend(y2.cpu().numpy().tolist()); y2_pred.extend(p2.tolist())
        leaf_true.extend(y_leaf_global.numpy().tolist()); leaf_pred.extend(p_leaf.tolist())

    metrics = {
        "loss": running_loss / max(1, len(loader)),
        "acc_level1": accuracy_score(y1_true, y1_pred),
        "acc_level2": accuracy_score(y2_true, y2_pred),
        "acc_leaf": accuracy_score(leaf_true, leaf_pred),
    }
    return metrics, np.array(y1_true), np.array(y1_pred), np.array(y2_true), np.array(y2_pred), np.array(leaf_true), np.array(leaf_pred)


# ============================================================
# SECTION 5 — DATA SETUP & MODEL INITIALIZATION
# ============================================================
# TRAIN/VAL/TEST PIPELINES:
# - TRAIN uses train_tfms_hier (augmentation + normalize)
# - VAL/TEST uses test_tfms_hier (deterministic + normalize)
#
# IMPORTANT FOR TWO-MODEL CONNECTION:
# - We save mean/std as mean1/std1
# - We save transforms as train_tfms_hier/test_tfms_hier
# ============================================================

train_dir = os.path.join(cfg.root_dir, "train")
val_dir   = os.path.join(cfg.root_dir, "val")
test_dir  = os.path.join(cfg.root_dir, "test")

mean1, std1 = compute_train_mean_std(train_dir, cfg.img_size, cfg.batch_size, cfg.num_workers)
print(f"\nComputed TRAIN mean (Model 1): {mean1}")
print(f"Computed TRAIN std  (Model 1): {std1}")

# Connection-safe transform names (won't be overwritten by Model 2)
train_tfms_hier, test_tfms_hier = build_transforms(mean1, std1)

train_ds = HierarchicalFoodDataset(train_dir, transform=train_tfms_hier, normalize_labels=cfg.normalize_labels)
val_ds   = HierarchicalFoodDataset(val_dir,   transform=test_tfms_hier,  normalize_labels=cfg.normalize_labels)
test_ds  = HierarchicalFoodDataset(test_dir,  transform=test_tfms_hier,  normalize_labels=cfg.normalize_labels)

train_loader = DataLoader(train_ds, batch_size=cfg.batch_size, shuffle=True,  num_workers=cfg.num_workers, pin_memory=(device.type=="cuda"))
val_loader   = DataLoader(val_ds,   batch_size=cfg.batch_size, shuffle=False, num_workers=cfg.num_workers, pin_memory=(device.type=="cuda"))
test_loader  = DataLoader(test_ds,  batch_size=cfg.batch_size, shuffle=False, num_workers=cfg.num_workers, pin_memory=(device.type=="cuda"))

print(f"\nTrain images: {len(train_ds)} | Val images: {len(val_ds)} | Test images: {len(test_ds)}")
print("Leaf classes:", LEAF_GLOBAL)

model = HierarchicalCNN().to(device)
criterion = nn.CrossEntropyLoss()

optimizer = optim.AdamW(model.parameters(), lr=cfg.lr, weight_decay=cfg.weight_decay)
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=cfg.epochs, eta_min=cfg.min_lr)


# ============================================================
# SECTION 6 — MODEL TRAINING & CHECKPOINT SELECTION
# ============================================================

def fmt_metric(x: float, digits_raw: int = 4, digits_pct: int = 2) -> str:
    """Pretty print metric as raw + percentage."""
    return f"{x:.{digits_raw}f} ({x*100:.{digits_pct}f}%)"


history = {
    "train_loss": [],
    "val_loss": [],
    "train_acc_leaf": [],
    "val_acc_leaf": [],
    "val_acc_level1": [],
    "val_acc_level2": [],
}

best_val_leaf = -1.0
best_epoch = -1
best_path = os.path.join(cfg.out_dir, "best_hierarchical_model.pth")

for ep in range(1, cfg.epochs + 1):
    train_loss = train_one_epoch(model, train_loader, optimizer, criterion, ep, cfg.epochs)
    train_metrics, *_ = evaluate(model, train_loader, criterion)
    val_metrics, *_ = evaluate(model, val_loader, criterion)

    scheduler.step()

    history["train_loss"].append(train_loss)
    history["val_loss"].append(val_metrics["loss"])
    history["train_acc_leaf"].append(train_metrics["acc_leaf"])
    history["val_acc_leaf"].append(val_metrics["acc_leaf"])
    history["val_acc_level1"].append(val_metrics["acc_level1"])
    history["val_acc_level2"].append(val_metrics["acc_level2"])

    print(
        f"[Epoch {ep}/{cfg.epochs}] "
        f"train_loss={train_loss:.4f} | val_loss={val_metrics['loss']:.4f} | "
        f"train_leaf={fmt_metric(train_metrics['acc_leaf'])} | "
        f"val_leaf={fmt_metric(val_metrics['acc_leaf'])} | "
        f"val_L1={fmt_metric(val_metrics['acc_level1'])} | "
        f"val_L2={fmt_metric(val_metrics['acc_level2'])} | "
        f"lr={optimizer.param_groups[0]['lr']:.2e}"
    )

    # Select best checkpoint by validation leaf accuracy
    if val_metrics["acc_leaf"] > best_val_leaf:
        best_val_leaf = val_metrics["acc_leaf"]
        best_epoch = ep
        torch.save(model.state_dict(), best_path)
        print(f"✅ Saved new best model at epoch {ep} (val_leaf={best_val_leaf*100:.2f}%)")

print(f"\nBest validation leaf accuracy: {fmt_metric(best_val_leaf)} (epoch {best_epoch})")
print(f"Saved best model to: {best_path}")


# ============================================================
# SECTION 7 — LEARNING CURVES & TEST SET EVALUATION
# ============================================================

def plot_learning_curves(history: Dict[str, List[float]]):
    """
    Plots:
      - Train loss vs Val loss
      - Train leaf accuracy vs Val leaf accuracy
    """
    epochs = list(range(1, len(history["train_loss"]) + 1))

    plt.figure()
    plt.plot(epochs, history["train_loss"], label="train_loss")
    plt.plot(epochs, history["val_loss"], label="val_loss")
    plt.xlabel("Epoch"); plt.ylabel("Loss"); plt.title("Loss vs Epoch"); plt.legend()
    plt.show()

    plt.figure()
    plt.plot(epochs, history["train_acc_leaf"], label="train_leaf_acc")
    plt.plot(epochs, history["val_acc_leaf"], label="val_leaf_acc")
    plt.xlabel("Epoch"); plt.ylabel("Accuracy"); plt.title("Leaf Accuracy vs Epoch"); plt.legend()
    plt.show()


plot_learning_curves(history)

# Load best checkpoint and evaluate on test split
model.load_state_dict(torch.load(best_path, map_location=device))
test_metrics, y1_true, y1_pred, y2_true, y2_pred, leaf_true, leaf_pred = evaluate(model, test_loader, criterion)

print("\nTEST RESULTS")
print(f"test_loss={test_metrics['loss']:.4f}")
print(f"test_acc_level1={fmt_metric(test_metrics['acc_level1'])}")
print(f"test_acc_level2={fmt_metric(test_metrics['acc_level2'])}")
print(f"test_acc_leaf={fmt_metric(test_metrics['acc_leaf'])}")


# ============================================================
# SECTION 8 — ERROR ANALYSIS & QUALITATIVE RESULTS
# ============================================================
# WHAT THIS SECTION DOES:
# - Confusion matrices and classification report for debugging
# - Shows a grid of random examples with True vs Pred hierarchy paths
# - Uses denormalize(mean1,std1) so images are displayed in natural colors


def plot_confusion_matrix_visual(cm, labels, title, normalize=False):
    """
    If normalize=True, shows row-normalized percentages (per true class).
    """
    cm_plot = cm.astype(np.float64)
    if normalize:
        row_sums = cm_plot.sum(axis=1, keepdims=True)
        cm_plot = np.divide(cm_plot, row_sums, out=np.zeros_like(cm_plot), where=row_sums != 0)

    fig, ax = plt.subplots(figsize=(max(8, 0.6 * len(labels)), max(6, 0.6 * len(labels))))
    im = ax.imshow(cm_plot, interpolation="nearest", cmap="Blues")
    ax.set_title(title)
    fig.colorbar(im, ax=ax)

    ax.set_xticks(np.arange(len(labels))); ax.set_yticks(np.arange(len(labels)))
    ax.set_xticklabels(labels, rotation=45, ha="right")
    ax.set_yticklabels(labels)
    ax.set_ylabel("True"); ax.set_xlabel("Pred")

    thresh = cm_plot.max() / 2.0 if cm_plot.size else 0.0
    for i in range(cm_plot.shape[0]):
        for j in range(cm_plot.shape[1]):
            txt = f"{cm_plot[i, j]*100:.1f}%" if normalize else f"{int(cm[i, j])}"
            ax.text(j, i, txt, ha="center", va="center",
                    color="white" if cm_plot[i, j] > thresh else "black", fontsize=8)

    plt.tight_layout()
    plt.show()


def denormalize(img_tensor, mean, std):
    """Invert Normalize(mean,std) for visualization."""
    mean_t = torch.tensor(mean).view(3, 1, 1)
    std_t = torch.tensor(std).view(3, 1, 1)
    return img_tensor * std_t + mean_t


@torch.no_grad()
def show_predictions_grid(model, dataset, mean, std, k=12, cols=4, title="Predictions (True vs Pred)"):
    """
    Shows random samples with:
      - True hierarchy folder path
      - Predicted hierarchy folder path
    """
    model.eval()
    k = min(k, len(dataset))
    idxs = torch.randperm(len(dataset))[:k].tolist()
    rows = math.ceil(k / cols)

    fig = plt.figure(figsize=(4 * cols, 4 * rows))
    plt.suptitle(title, fontsize=14)

    for i, idx in enumerate(idxs):
        x, _, _, _, _, _, meta = dataset[idx]
        xb = x.unsqueeze(0).to(device)
        _, _, _, pred_strings = predict_hierarchy(model, xb)

        pred_lvl1, pred_lvl2, pred_leaf = pred_strings[0]
        true_path = f"{meta['lvl1']} / {meta['lvl2']} / {meta['leaf']}"
        pred_path = f"{pred_lvl1} / {pred_lvl2} / {pred_leaf}"

        x_vis = denormalize(x.cpu(), mean, std).clamp(0, 1).permute(1, 2, 0).numpy()

        ax = plt.subplot(rows, cols, i + 1)
        ax.imshow(x_vis); ax.axis("off")
        ax.set_title(f"T: {true_path}\nP: {pred_path}", fontsize=9)

    plt.tight_layout(rect=[0, 0, 1, 0.95])
    plt.show()
    plt.close(fig)


# ---- Confusion matrices ----
cm_leaf = confusion_matrix(leaf_true, leaf_pred, labels=list(range(len(LEAF_GLOBAL))))
cm_lvl1 = confusion_matrix(y1_true, y1_pred, labels=[0, 1])
cm_lvl2 = confusion_matrix(y2_true, y2_pred, labels=[0, 1])

plot_confusion_matrix_visual(cm_leaf, LEAF_GLOBAL, "Leaf Confusion Matrix (Counts)", normalize=False)
plot_confusion_matrix_visual(cm_leaf, LEAF_GLOBAL, "Leaf Confusion Matrix (Row %)", normalize=True)
plot_confusion_matrix_visual(cm_lvl1, LEVEL1, "Level 1 Confusion Matrix (Counts)", normalize=False)
plot_confusion_matrix_visual(cm_lvl1, LEVEL1, "Level 1 Confusion Matrix (Row %)", normalize=True)
plot_confusion_matrix_visual(cm_lvl2, ["Group0", "Group1"], "Level 2 Confusion Matrix (Counts)", normalize=False)
plot_confusion_matrix_visual(cm_lvl2, ["Group0", "Group1"], "Level 2 Confusion Matrix (Row %)", normalize=True)

# ---- Per-class precision/recall/F1 report (Leaf level) ----
print("\nCLASSIFICATION REPORT (Leaf level):")
print(classification_report(leaf_true, leaf_pred, target_names=LEAF_GLOBAL, digits=4))

# ---- Qualitative samples ----
show_predictions_grid(model, test_ds, mean1, std1, k=cfg.num_example_preds, cols=4)


# ============================================================
# FINAL STEP — CONNECTION-SAFE EXPORTS FOR MODEL 2 PIPELINE
# ============================================================
# This prevents the second model from overwriting variables when both are in the same notebook.
# ============================================================

model_hier = model
model_hier.eval()

print("\n✅ Model 1 ready for combined pipeline:")
print("   - model_hier      (trained hierarchical model)")
print("   - mean1/std1      (normalization stats for Model 1)")
print("   - test_tfms_hier  (deterministic eval transform for inference)")
```

```python
# ============================================================
# PER-CLASS (LEAF) ACCURACY — TEST SET
# ============================================================
# PURPOSE:
#   - Compute accuracy separately for each leaf class.
#   - This answers: "When the true class is X, how often is X predicted correctly?"
#   - Useful for identifying weak classes even when overall accuracy is high.
# ============================================================

def print_leaf_class_accuracy(y_true, y_pred, class_names):
    """
    Prints per-class accuracy (%).

    Accuracy for class c is defined as:
        (# correctly predicted samples of class c) / (total samples of class c)

    Parameters
    ----------
    y_true : np.ndarray
        Ground-truth leaf labels (global index space).
    y_pred : np.ndarray
        Predicted leaf labels (global index space).
    class_names : List[str]
        Leaf class names corresponding to label indices.
    """
    print("\nPER-LEAF CLASS ACCURACY (TEST SET):")
    print("-" * 50)

    for class_idx, class_name in enumerate(class_names):
        mask = (y_true == class_idx)
        total = mask.sum()

        if total == 0:
            acc = float("nan")
            print(f"{class_name:15s} : N/A (no samples)")
        else:
            correct = (y_pred[mask] == class_idx).sum()
            acc = 100.0 * correct / total
            print(f"{class_name:15s} : {acc:6.2f}%  ({correct}/{total})")


# Call after test evaluation
print_leaf_class_accuracy(
    y_true=leaf_true,
    y_pred=leaf_pred,
    class_names=LEAF_GLOBAL
)
```

```python
# ============================================================
# CNN MODEL 2 - Singular Multi-Head 
# ============================================================c
# ============================================================
# FULL SCRIPT (Sections 1–8) — CLEAN + RUNNABLE + OFFLINE CROP (OTSU BBOX)
#
# Key idea:
#   - Use Otsu ONLY to find an object mask -> bounding box -> CROP (with padding).
#   - No whitening, no pixel removal, no attention blending.
#   - Build a new dataset folder ONCE (offline), then train on it.
#
# Why this is better than masking/attention here:
#   - Your background is already clean; attention barely changes images.
#   - Hard masking can introduce holes / distortions.
#   - Cropping actually changes the data distribution (less background),
#     while keeping the fruit pixels intact (good for condition classification).
# ============================================================

# ============================================================
# SECTION 1 — Environment Setup & Experiment Configuration
# ============================================================
import os
import random
import math
from dataclasses import dataclass
from typing import List, Dict, Tuple, Optional
from pathlib import Path
from collections import defaultdict

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from tqdm import tqdm

from sklearn.metrics import confusion_matrix, classification_report, accuracy_score

# --- scikit-image (lecture-aligned CV) ---
from skimage.feature import canny
from skimage.filters import threshold_otsu
from skimage.morphology import remove_small_objects, remove_small_holes
from skimage.measure import label, regionprops
from skimage.morphology import square
from skimage.filters.rank import median as rank_median

import warnings
warnings.filterwarnings(
    "ignore",
    message="Palette images with Transparency expressed in bytes should be converted to RGBA images"
)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")


@dataclass
class Config:
    # Original dataset with train/val/test
    root_dir: str = r"C:\Users\Yioulika Antoniades\Desktop\UDA_Datasets\Dataset2_FruitCond"  # <-- CHANGE THIS ONLY

    # Offline-preprocessed dataset output (CROPPED dataset)
    preprocess_out_dir: str = r"C:\Users\Yioulika Antoniades\Desktop\UDA_Datasets\Dataset2_FruitCond_CROP"

    # Train on offline cropped dataset?
    use_offline_crop: bool = True

    # Build offline cropped dataset in this run?
    # Set True only when you want to create/update the offline dataset.
    build_offline_crop_now: bool = True

    img_size: int = 160
    batch_size: int = 8
    num_workers: int = 0
    epochs: int = 25

    # Optimizer (SGD)
    lr: float = 0.01
    momentum: float = 0.9
    weight_decay: float = 1e-4

    # Scheduler
    step_size: int = 15
    gamma: float = 0.3

    # Outputs
    out_dir: str = "outputs"
    best_model_name: str = "best_fruit_condition_cnn.pt"
    num_example_preds: int = 12

    seed: int = 42
    normalize_labels: bool = True

    # Classical CV augmentation probabilities (TRAIN ONLY)
    p_median: float = 0.20
    p_edges: float = 0.20

    # Canny sigmas (multi-scale)
    canny_sigmas: Tuple[float, ...] = (1.0, 3.0)

    # Offline crop padding around bbox
    crop_pad: int = 14

    # Section 3 outputs
    viz_k_examples: int = 6
    viz_seed: int = 42


cfg = Config()


def seed_everything(seed: int = 42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


seed_everything(cfg.seed)

# ============================================================
# SECTION 2 — Label Hierarchy & Dataset Integrity Check
# ============================================================
FRUITS = ["Apple", "Banana", "Grape", "Mango"]
CONDITIONS = ["Fresh", "Rotten", "Formalin-mixed"]

IMAGE_EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".webp", ".tiff", ".gif"}


def _norm(s: str) -> str:
    return " ".join(s.strip().lower().split())


def _maybe_norm(s: str, enabled: bool) -> str:
    return _norm(s) if enabled else s


cond_to_idx = {_maybe_norm(c, cfg.normalize_labels): i for i, c in enumerate(CONDITIONS)}
idx_to_cond = {i: c for i, c in enumerate(CONDITIONS)}


def list_images_pathlib(folder: Path) -> List[Path]:
    """Recursively list all valid image files under a folder."""
    if not folder.exists():
        return []
    return [p for p in folder.rglob("*") if p.is_file() and p.suffix.lower() in IMAGE_EXTS]


def folder_size_bytes(folder: Path) -> int:
    """Total disk size (bytes) of all files under a folder."""
    if not folder.exists():
        return 0
    total = 0
    for p in folder.rglob("*"):
        if p.is_file():
            try:
                total += p.stat().st_size
            except OSError:
                pass
    return total


def human_bytes(num_bytes: int) -> str:
    """Convert bytes → human-readable string."""
    units = ["B", "KB", "MB", "GB", "TB"]
    size = float(num_bytes)
    for u in units:
        if size < 1024 or u == units[-1]:
            return f"{size:.2f} {u}"
        size /= 1024


def leaf_class_counts(split_dir: Path):
    """
    Dataset structure:
        split/Fruit/Condition/image
    Leaf class = Fruit/Condition
    """
    images = list_images_pathlib(split_dir)
    counts = defaultdict(int)
    for img in images:
        counts[img.parent] += 1
    rel_counts = {str(folder.relative_to(split_dir)): n for folder, n in counts.items()}
    return rel_counts, len(images)


def inspect_dataset(root_dir: str):
    root = Path(root_dir)
    splits = {"train": root / "train", "val": root / "val", "test": root / "test"}

    print("\nDataset successfully loaded from local directory.")
    print(f"Dataset root path: {root}")

    missing = [k for k, p in splits.items() if not p.exists()]
    if missing:
        print("\nWARNING: Some expected split folders were not found:")
        for k in missing:
            print(f" - Missing: {k} -> {splits[k]}")
        print("\nExpected structure:")
        print("  root/train/<Fruit>/<Condition>/image")
        print("  root/val/<Fruit>/<Condition>/image")
        print("  root/test/<Fruit>/<Condition>/image")
        return False

    total_bytes = folder_size_bytes(root)
    print(f"\nTotal dataset size on disk: {human_bytes(total_bytes)}")

    grand_total_images = 0
    leaf_folders_union = set()

    print("\nDataset structure (split → Fruit/Condition folders with image counts):")
    for split_name, split_dir in splits.items():
        rel_counts, split_total = leaf_class_counts(split_dir)
        print(f"\n- {split_name.upper()} split:")
        print(f"  Path: {split_dir}")
        print(f"  Images in {split_name}: {split_total}")
        for rel_path in sorted(rel_counts.keys()):
            print(f"    - {rel_path} : {rel_counts[rel_path]} images")

        grand_total_images += split_total
        leaf_folders_union.update(rel_counts.keys())

    print("\nSummary:")
    print(f"  Number of splits: {len(splits)}")
    print(f"  Unique Fruit/Condition folders: {len(leaf_folders_union)}")
    print(f"  Total number of images (train + val + test): {grand_total_images}")
    return True


_ = inspect_dataset(cfg.root_dir)

# ============================================================
# SECTION 3 — Offline Crop Preprocessing + Train Augmentations
#
# Outputs when runner executes:
#   1) Build stats (Created/Skipped/Failed)
#   2) Confirm new dataset exists + image counts raw vs crop
#   3) Print sample output paths
#   4) Show samples from the NEW dataset
#   5) Show RAW vs OTSU MASK vs CROPPED examples
# ============================================================

# ---------------------------
# Image helpers
# ---------------------------
def pil_to_np_rgb(pil_img: Image.Image) -> np.ndarray:
    return np.array(pil_img.convert("RGB"), dtype=np.uint8)


def np_rgb_to_pil(arr: np.ndarray) -> Image.Image:
    arr = np.clip(arr, 0, 255).astype(np.uint8)
    return Image.fromarray(arr, mode="RGB")


def to_gray(rgb: np.ndarray) -> np.ndarray:
    # Only for mask computation; training remains RGB
    return (0.299 * rgb[..., 0] + 0.587 * rgb[..., 1] + 0.114 * rgb[..., 2]).astype(np.uint8)


# ---------------------------
# TRAIN augmentation ops (lecture aligned)
# ---------------------------
def median_denoise(rgb: np.ndarray, k: int = 3) -> np.ndarray:
    out = np.empty_like(rgb)
    fp = square(k)
    for c in range(3):
        out[..., c] = rank_median(rgb[..., c], footprint=fp)
    return out


def canny_edges_scikit(gray_u8: np.ndarray, sigma: float = 1.0) -> np.ndarray:
    gray_norm = gray_u8.astype(np.float32) / 255.0
    return canny(gray_norm, sigma=sigma)


def overlay_edges(rgb: np.ndarray, edges: np.ndarray, alpha: float = 0.75) -> np.ndarray:
    out = rgb.astype(np.float32).copy()
    out[edges] = out[edges] * (1.0 - alpha)
    return np.clip(out, 0, 255).astype(np.uint8)


class ClassicalFruitAugmentScikit:
    """
    TRAIN-only lightweight augmentation:
      (1) Median denoise (optional)
      (2) Canny edge overlay (optional)
    """
    def __init__(self, p_median: float, p_edges: float, canny_sigmas: Tuple[float, ...], seed: int = 42):
        self.p_median = float(p_median)
        self.p_edges = float(p_edges)
        self.canny_sigmas = tuple(canny_sigmas)
        self.rng = random.Random(seed)

    def __call__(self, img: Image.Image) -> Image.Image:
        rgb = pil_to_np_rgb(img)

        if self.rng.random() < self.p_median:
            rgb = median_denoise(rgb, k=3)

        if self.rng.random() < self.p_edges:
            sigma = self.rng.choice(list(self.canny_sigmas))
            edges = canny_edges_scikit(to_gray(rgb), sigma=sigma)
            rgb = overlay_edges(rgb, edges, alpha=0.75)

        return np_rgb_to_pil(rgb)


# ---------------------------
# Otsu mask -> bbox crop (offline)
# ---------------------------
def largest_connected_component(mask: np.ndarray) -> np.ndarray:
    if mask.sum() == 0:
        return mask
    lab = label(mask)
    props = regionprops(lab)
    if not props:
        return mask
    largest = max(props, key=lambda r: r.area).label
    return lab == largest


def otsu_foreground_mask(gray_u8: np.ndarray) -> np.ndarray:
    """
    Robust Otsu:
      - try <= and >=
      - clean + keep largest component
      - choose the one with larger main component
    """
    t = threshold_otsu(gray_u8)

    def clean(m):
        m = remove_small_holes(m, area_threshold=800)
        m = remove_small_objects(m, min_size=800)
        m = largest_connected_component(m)
        return m

    m1 = clean(gray_u8 <= t)
    m2 = clean(gray_u8 >= t)
    return m1 if m1.sum() >= m2.sum() else m2


def bbox_from_mask(mask: np.ndarray, pad: int) -> Optional[Tuple[int, int, int, int]]:
    ys, xs = np.where(mask)
    if len(xs) == 0:
        return None

    x1, x2 = xs.min(), xs.max()
    y1, y2 = ys.min(), ys.max()

    x1 = max(0, x1 - pad)
    y1 = max(0, y1 - pad)
    x2 = min(mask.shape[1] - 1, x2 + pad)
    y2 = min(mask.shape[0] - 1, y2 + pad)
    return x1, y1, x2, y2


def offline_crop_one_image(pil_img: Image.Image, pad: int) -> Tuple[Image.Image, np.ndarray]:
    """
    Returns:
      - cropped PIL image (RGB)
      - Otsu mask for visualization/debug
    """
    rgb = pil_to_np_rgb(pil_img)
    gray = to_gray(rgb)
    mask = otsu_foreground_mask(gray)

    bb = bbox_from_mask(mask, pad=pad)
    if bb is None:
        # if mask fails, just return original (safe fallback)
        return pil_img.convert("RGB"), mask

    x1, y1, x2, y2 = bb
    cropped = pil_img.convert("RGB").crop((x1, y1, x2 + 1, y2 + 1))
    return cropped, mask


def build_offline_cropped_dataset(in_root: str, out_root: str, pad: int, image_exts=None):
    """
    Create out_root dataset with identical structure to in_root.
    Each image is cropped using Otsu mask bbox + padding.
    Skips existing files (resume-safe).
    """
    if image_exts is None:
        image_exts = IMAGE_EXTS

    in_root = Path(in_root)
    out_root = Path(out_root)
    out_root.mkdir(parents=True, exist_ok=True)

    splits = ["train", "val", "test"]
    created, skipped, failed = 0, 0, 0
    sample_outputs: List[Path] = []

    print(f"\n[OFFLINE-CROP] Input : {in_root}")
    print(f"[OFFLINE-CROP] Output: {out_root}")
    print(f"[OFFLINE-CROP] pad={pad}")

    for split in splits:
        in_split = in_root / split
        out_split = out_root / split
        out_split.mkdir(parents=True, exist_ok=True)

        if not in_split.exists():
            print(f"[OFFLINE-CROP] Missing split folder: {in_split} (skipping)")
            continue

        for fruit_dir in in_split.iterdir():
            if not fruit_dir.is_dir():
                continue

            for cond_dir in fruit_dir.iterdir():
                if not cond_dir.is_dir():
                    continue

                out_cond = out_split / fruit_dir.name / cond_dir.name
                out_cond.mkdir(parents=True, exist_ok=True)

                for img_path in cond_dir.iterdir():
                    if not (img_path.is_file() and img_path.suffix.lower() in image_exts):
                        continue

                    out_path = out_cond / img_path.name

                    if out_path.exists():
                        skipped += 1
                        continue

                    try:
                        img = Image.open(img_path).convert("RGB")
                        cropped_img, _ = offline_crop_one_image(img, pad=pad)

                        try:
                            cropped_img.save(out_path)
                        except Exception:
                            out_path = out_path.with_suffix(".png")
                            cropped_img.save(out_path)

                        created += 1
                        if len(sample_outputs) < 8:
                            sample_outputs.append(out_path)

                    except Exception as e:
                        failed += 1
                        print(f"[OFFLINE-CROP] Failed: {img_path} -> {e}")

    print(f"\n[OFFLINE-CROP] Done. Created={created}, Skipped(existing)={skipped}, Failed={failed}")
    if sample_outputs:
        print("[OFFLINE-CROP] Sample output paths:")
        for p in sample_outputs:
            print("  -", p)

    return {"created": created, "skipped": skipped, "failed": failed}


def _collect_images(root: Path, image_exts: set) -> List[Path]:
    return [p for p in root.rglob("*") if p.is_file() and p.suffix.lower() in image_exts]


def verify_crop_dataset_created(raw_root: str, crop_root: str, sample_n: int = 10):
    raw_root = Path(raw_root)
    crop_root = Path(crop_root)

    print("\n[VERIFY] Raw root :", raw_root)
    print("[VERIFY] Crop root:", crop_root)
    print("[VERIFY] Exists?  :", crop_root.exists())

    if not crop_root.exists():
        return

    raw_imgs = _collect_images(raw_root, IMAGE_EXTS)
    crop_imgs = _collect_images(crop_root, IMAGE_EXTS)

    print(f"[VERIFY] RAW images : {len(raw_imgs)}")
    print(f"[VERIFY] CROP images: {len(crop_imgs)}")

    if crop_imgs:
        print("\n[VERIFY] Sample cropped files:")
        for p in random.sample(crop_imgs, k=min(sample_n, len(crop_imgs))):
            print(" -", p)


def show_crop_dataset_samples(crop_root: str, split: str = "train", k: int = 12, cols: int = 4, seed: int = 42):
    rng = random.Random(seed)
    crop_split = Path(crop_root) / split
    imgs = [p for p in crop_split.rglob("*") if p.is_file() and p.suffix.lower() in IMAGE_EXTS]

    if len(imgs) == 0:
        print("No cropped images found to display.")
        return

    samples = rng.sample(imgs, k=min(k, len(imgs)))
    rows = math.ceil(len(samples) / cols)

    fig = plt.figure(figsize=(4 * cols, 4 * rows))
    plt.suptitle(f"CROPPED DATASET SAMPLES ({split.upper()})", fontsize=14)

    for i, p in enumerate(samples):
        img = Image.open(p).convert("RGB")
        ax = plt.subplot(rows, cols, i + 1)
        ax.imshow(img)
        ax.axis("off")
        ax.set_title(str(p.relative_to(crop_split))[:35], fontsize=8)

    plt.tight_layout(rect=[0, 0, 1, 0.95])
    plt.show()


def show_raw_mask_crop_examples(raw_root: str, crop_root: str, split: str = "train", k: int = 6, seed: int = 42):
    rng = random.Random(seed)
    raw_split = Path(raw_root) / split
    crop_split = Path(crop_root) / split

    raw_images = [p for p in raw_split.rglob("*") if p.is_file() and p.suffix.lower() in IMAGE_EXTS]
    if len(raw_images) == 0:
        print("No images found for visualization.")
        return

    samples = rng.sample(raw_images, min(k, len(raw_images)))
    fig, axes = plt.subplots(len(samples), 3, figsize=(12, 4 * len(samples)))
    fig.suptitle(f"RAW vs OTSU MASK vs CROPPED ({split.upper()})", fontsize=14)

    if len(samples) == 1:
        axes = np.expand_dims(axes, axis=0)

    for i, raw_path in enumerate(samples):
        rel = raw_path.relative_to(raw_split)
        crop_path = crop_split / rel

        raw_img = Image.open(raw_path).convert("RGB")
        raw_rgb = pil_to_np_rgb(raw_img)
        mask = otsu_foreground_mask(to_gray(raw_rgb))

        if crop_path.exists():
            crop_img = Image.open(crop_path).convert("RGB")
        else:
            crop_img, _ = offline_crop_one_image(raw_img, pad=cfg.crop_pad)

        axes[i, 0].imshow(raw_img); axes[i, 0].set_title("RAW"); axes[i, 0].axis("off")
        axes[i, 1].imshow(mask, cmap="gray"); axes[i, 1].set_title("OTSU MASK"); axes[i, 1].axis("off")
        axes[i, 2].imshow(crop_img); axes[i, 2].set_title(f"CROP (pad={cfg.crop_pad})"); axes[i, 2].axis("off")

    plt.tight_layout(rect=[0, 0, 1, 0.95])
    plt.show()


# ---------------------------
# SECTION 3 RUNNER
# ---------------------------
if cfg.build_offline_crop_now:
    stats = build_offline_cropped_dataset(
        in_root=cfg.root_dir,
        out_root=cfg.preprocess_out_dir,
        pad=cfg.crop_pad,
        image_exts=IMAGE_EXTS
    )
else:
    print("\n[OFFLINE-CROP] Skipped building (cfg.build_offline_crop_now=False)")
    stats = {"created": 0, "skipped": 0, "failed": 0}

verify_crop_dataset_created(cfg.root_dir, cfg.preprocess_out_dir, sample_n=10)
show_crop_dataset_samples(cfg.preprocess_out_dir, split="train", k=12, cols=4, seed=cfg.seed)
show_raw_mask_crop_examples(cfg.root_dir, cfg.preprocess_out_dir, split="train", k=cfg.viz_k_examples, seed=cfg.viz_seed)


# ============================================================
# Dataset class
# ============================================================
class FruitConditionDataset(Dataset):
    """
    Reads: split/Fruit/Condition/image
    Label = Condition only (Fresh/Rotten/Formalin-mixed)
    """
    def __init__(self, split_dir: str, transform=None):
        self.split_dir = Path(split_dir)
        self.transform = transform
        self.samples: List[Tuple[Path, int]] = []

        if not self.split_dir.exists():
            raise FileNotFoundError(f"Split folder not found: {self.split_dir}")

        for fruit_dir in self.split_dir.iterdir():
            if not fruit_dir.is_dir():
                continue
            if fruit_dir.name not in FRUITS:
                continue

            for cond_dir in fruit_dir.iterdir():
                if not cond_dir.is_dir():
                    continue

                cond_key = _maybe_norm(cond_dir.name, cfg.normalize_labels)
                if cond_key not in cond_to_idx:
                    continue

                y = cond_to_idx[cond_key]
                for img_path in cond_dir.iterdir():
                    if img_path.is_file() and img_path.suffix.lower() in IMAGE_EXTS:
                        self.samples.append((img_path, y))

        if len(self.samples) == 0:
            raise RuntimeError(
                f"No images found in {self.split_dir}\n"
                "Expected: split/<Fruit>/<Fresh|Rotten|Formalin-mixed>/*"
            )

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx: int):
        img_path, y = self.samples[idx]
        img = Image.open(img_path).convert("RGB")
        if self.transform:
            img = self.transform(img)
        return img, y


# ============================================================
# SECTION 4 — Model Architecture & Core Training/ Evaluation Logic
# ============================================================
class Net(nn.Module):
    def __init__(self, num_classes: int = 3):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 16, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            nn.Conv2d(16, 64, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            nn.Conv2d(64, 128, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),
        )
        self.gap = nn.AdaptiveAvgPool2d((4, 4))
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(128 * 4 * 4, 256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
            nn.Linear(256, num_classes),
        )

    def forward(self, x):
        x = self.features(x)
        x = self.gap(x)
        x = self.classifier(x)
        return x


def train_one_epoch(model, loader, optimizer, criterion, epoch_idx, total_epochs):
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0

    pbar = tqdm(loader, desc=f"Epoch {epoch_idx}/{total_epochs} [train]", unit="batch")
    for images, labels in pbar:
        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()
        preds = outputs.argmax(dim=1)
        total += labels.size(0)
        correct += (preds == labels).sum().item()
        pbar.set_postfix(loss=f"{loss.item():.4f}")

    avg_loss = running_loss / max(1, len(loader))
    acc = correct / max(1, total)
    return avg_loss, acc


@torch.no_grad()
def evaluate(model, loader, criterion):
    model.eval()
    running_loss = 0.0
    y_true, y_pred = [], []

    for images, labels in loader:
        images = images.to(device)
        labels = labels.to(device)

        outputs = model(images)
        loss = criterion(outputs, labels)
        running_loss += loss.item()

        preds = outputs.argmax(dim=1)
        y_true.extend(labels.cpu().tolist())
        y_pred.extend(preds.cpu().tolist())

    acc = accuracy_score(y_true, y_pred)
    metrics = {"loss": running_loss / max(1, len(loader)), "acc": acc}
    return metrics, y_true, y_pred


@torch.no_grad()
def compute_mean_std_from_train(train_dir: str, img_size: int, max_images: int = 2000) -> Tuple[List[float], List[float]]:
    tf = transforms.Compose([
        transforms.Resize((img_size, img_size)),
        transforms.ToTensor(),
    ])
    ds = FruitConditionDataset(train_dir, transform=tf)
    n = min(len(ds), max_images)
    idxs = torch.randperm(len(ds))[:n].tolist()

    imgs = []
    for i in idxs:
        x, _ = ds[i]
        imgs.append(x)

    stack = torch.stack(imgs, dim=0)
    mean = stack.mean(dim=[0, 2, 3]).tolist()
    std = stack.std(dim=[0, 2, 3]).tolist()
    return mean, std


# ============================================================
# SECTION 5 — Data Setup & Model Initialisation
# ============================================================
def build_loaders(root_dir: str, img_size: int, batch_size: int, num_workers: int, mean: List[float], std: List[float]):
    aug = ClassicalFruitAugmentScikit(
        p_median=cfg.p_median,
        p_edges=cfg.p_edges,
        canny_sigmas=cfg.canny_sigmas,
        seed=cfg.seed
    )

    train_tf = transforms.Compose([
        aug,
        transforms.RandomResizedCrop(img_size),
        transforms.RandomHorizontalFlip(),
        transforms.ToTensor(),
        transforms.Normalize(mean=mean, std=std),
    ])

    eval_tf = transforms.Compose([
        transforms.Resize(img_size + 16),
        transforms.CenterCrop(img_size),
        transforms.ToTensor(),
        transforms.Normalize(mean=mean, std=std),
    ])

    train_dir = os.path.join(root_dir, "train")
    val_dir   = os.path.join(root_dir, "val")
    test_dir  = os.path.join(root_dir, "test")

    train_ds = FruitConditionDataset(train_dir, transform=train_tf)
    val_ds   = FruitConditionDataset(val_dir, transform=eval_tf)
    test_ds  = FruitConditionDataset(test_dir, transform=eval_tf)

    train_loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True,  num_workers=num_workers, pin_memory=(device.type == "cuda"))
    val_loader   = DataLoader(val_ds,   batch_size=batch_size, shuffle=False, num_workers=num_workers, pin_memory=(device.type == "cuda"))
    test_loader  = DataLoader(test_ds,  batch_size=batch_size, shuffle=False, num_workers=num_workers, pin_memory=(device.type == "cuda"))
    return train_loader, val_loader, test_loader, train_ds, val_ds, test_ds


train_root = cfg.preprocess_out_dir if cfg.use_offline_crop else cfg.root_dir
print(f"\nTraining root selected: {train_root}")
print(f"Using OFFLINE cropped dataset? {cfg.use_offline_crop}")

train_dir = os.path.join(train_root, "train")
mean, std = compute_mean_std_from_train(train_dir, cfg.img_size, max_images=800)
print("\nComputed TRAIN mean:", mean)
print("Computed TRAIN std :", std)

train_loader, val_loader, test_loader, train_ds, val_ds, test_ds = build_loaders(
    train_root, cfg.img_size, cfg.batch_size, cfg.num_workers, mean, std
)
print(f"\nDataset sizes: train={len(train_ds)}, val={len(val_ds)}, test={len(test_ds)}")
print("Classes:", CONDITIONS)

model = Net(num_classes=3).to(device)
criterion = nn.CrossEntropyLoss()

optimizer = optim.SGD(
    model.parameters(),
    lr=cfg.lr,
    momentum=cfg.momentum,
    weight_decay=cfg.weight_decay
)
scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=cfg.step_size, gamma=cfg.gamma)

os.makedirs(cfg.out_dir, exist_ok=True)
best_path = os.path.join(cfg.out_dir, cfg.best_model_name)
print(f"Best model will be saved to: {best_path}")


# ============================================================
# SECTION 6 — Model Training
# ============================================================
history = {"train_loss": [], "val_loss": [], "train_acc": [], "val_acc": []}
best_val_acc = -1.0

for ep in range(1, cfg.epochs + 1):
    train_loss, train_acc = train_one_epoch(model, train_loader, optimizer, criterion, ep, cfg.epochs)
    val_metrics, _, _ = evaluate(model, val_loader, criterion)

    history["train_loss"].append(train_loss)
    history["val_loss"].append(val_metrics["loss"])
    history["train_acc"].append(train_acc * 100.0)
    history["val_acc"].append(val_metrics["acc"] * 100.0)

    print(f"[Epoch {ep}/{cfg.epochs}] "
          f"train_loss={train_loss:.4f} | val_loss={val_metrics['loss']:.4f} | "
          f"train_acc={train_acc*100:.2f}% | val_acc={val_metrics['acc']*100:.2f}%")

    if val_metrics["acc"] > best_val_acc:
        best_val_acc = val_metrics["acc"]
        torch.save(model.state_dict(), best_path)
        print(f"✅ Saved new best model: {best_path} (val_acc={best_val_acc*100:.2f}%)")

    scheduler.step()

print("\nFinished Training.")


# ============================================================
# SECTION 7 — Learning Curves & Test Set Evaluation
# ============================================================
def plot_learning_curves(history: Dict[str, List[float]]):
    epochs = list(range(1, len(history["train_loss"]) + 1))

    plt.figure()
    plt.plot(epochs, history["train_loss"], label="train_loss")
    plt.plot(epochs, history["val_loss"], label="val_loss")
    plt.xlabel("Epoch"); plt.ylabel("Loss"); plt.title("Loss vs Epoch"); plt.legend()
    plt.show()

    plt.figure()
    plt.plot(epochs, history["train_acc"], label="train_acc")
    plt.plot(epochs, history["val_acc"], label="val_acc")
    plt.xlabel("Epoch"); plt.ylabel("Accuracy (%)"); plt.title("Accuracy vs Epoch"); plt.legend()
    plt.show()


plot_learning_curves(history)

model.load_state_dict(torch.load(best_path, map_location=device))
test_metrics, y_true, y_pred = evaluate(model, test_loader, criterion)

print("\nTEST RESULTS")
print(f"test_loss={test_metrics['loss']:.4f}")
print(f"test_acc={test_metrics['acc']*100:.2f}%")


# ============================================================
# SECTION 8 — Error Analysis & Qualitative Results
# ============================================================
def plot_confusion_matrix_visual(cm, labels, title, normalize=False):
    cm_plot = cm.astype(np.float64)
    if normalize:
        row_sums = cm_plot.sum(axis=1, keepdims=True)
        cm_plot = np.divide(cm_plot, row_sums, out=np.zeros_like(cm_plot), where=row_sums != 0)

    fig, ax = plt.subplots(figsize=(8, 6))
    im = ax.imshow(cm_plot, interpolation="nearest", cmap="Blues")
    ax.set_title(title)
    fig.colorbar(im, ax=ax)

    ax.set_xticks(np.arange(len(labels))); ax.set_yticks(np.arange(len(labels)))
    ax.set_xticklabels(labels, rotation=45, ha="right")
    ax.set_yticklabels(labels)
    ax.set_ylabel("True"); ax.set_xlabel("Pred")

    thresh = cm_plot.max() / 2.0 if cm_plot.size else 0.0
    for i in range(cm_plot.shape[0]):
        for j in range(cm_plot.shape[1]):
            txt = f"{cm_plot[i, j]*100:.1f}%" if normalize else f"{int(cm[i, j])}"
            ax.text(j, i, txt, ha="center", va="center",
                    color="white" if cm_plot[i, j] > thresh else "black", fontsize=9)

    plt.tight_layout()
    plt.show()


def denormalize(img_tensor, mean, std):
    mean_t = torch.tensor(mean).view(3, 1, 1)
    std_t = torch.tensor(std).view(3, 1, 1)
    return img_tensor * std_t + mean_t


@torch.no_grad()
def show_predictions_grid(model, dataset, mean, std, k=12, cols=4, title="Predictions (True vs Pred)"):
    model.eval()
    k = min(k, len(dataset))
    idxs = torch.randperm(len(dataset))[:k].tolist()
    rows = math.ceil(k / cols)

    fig = plt.figure(figsize=(4 * cols, 4 * rows))
    plt.suptitle(title, fontsize=14)

    for i, idx in enumerate(idxs):
        x, y = dataset[idx]
        xb = x.unsqueeze(0).to(device)
        out = model(xb)
        pred = out.argmax(dim=1).item()

        x_vis = denormalize(x.cpu(), mean, std).clamp(0, 1).permute(1, 2, 0).numpy()

        ax = plt.subplot(rows, cols, i + 1)
        ax.imshow(x_vis); ax.axis("off")
        ax.set_title(f"T: {idx_to_cond[y]}\nP: {idx_to_cond[pred]}", fontsize=10)

    plt.tight_layout(rect=[0, 0, 1, 0.95])
    plt.show()
    plt.close(fig)


cm = confusion_matrix(y_true, y_pred, labels=list(range(len(CONDITIONS))))
plot_confusion_matrix_visual(cm, CONDITIONS, "Confusion Matrix (Counts)", normalize=False)
plot_confusion_matrix_visual(cm, CONDITIONS, "Confusion Matrix (Row %)", normalize=True)

print("\nCLASSIFICATION REPORT (Condition level):")
print(classification_report(y_true, y_pred, target_names=CONDITIONS, digits=4))

show_predictions_grid(model, test_ds, mean, std, k=cfg.num_example_preds, cols=4)
```

## Hierarchical–Conditional Food Inspection System

```python
# ============================================================
# SINGLE-IMAGE CASCADED INFERENCE (EXAMINER TEST) — SELF-CONTAINED
#
# PURPOSE:
#   Examiner runs ONE cell, edits only IMAGE_PATH (and the file paths
#   to the saved models/stats if needed), and gets the final output.
#
# PIPELINE:
#   1) Model 1 (Hierarchical CNN) runs first -> predicts leaf class (e.g., Apples, Fish, etc.)
#   2) If the predicted leaf is one of {Apple, Banana, Grape, Mango},
#      then Model 2 (Condition CNN) runs -> predicts {Fresh, Rotten, Formalin-mixed}
#   3) Otherwise Model 2 is skipped (prints a clear message)
#
# REQUIREMENTS:
#   - You must have already trained & saved:
#       (A) Model 1 weights file (e.g. best_hierarchical_model.pth)
#       (B) Model 1 normalization stats file (JSON: mean/std)
#       (C) Model 2 weights file (e.g. best_fruit_condition_cnn.pt)
#       (D) Model 2 normalization stats file (JSON: mean/std)
#
#   Why stats files?
#     So the examiner does NOT need to re-run training to obtain mean/std.
#
# WHAT THIS CELL DOES NOT DO:
#   - No training
#   - No dataset loading
# ============================================================

import os
import json
import math
from dataclasses import dataclass
from typing import List, Tuple, Dict

import numpy as np
import torch
import torch.nn as nn
from torchvision import transforms
from PIL import Image

# ---------------------------
# EXAMINER INPUT (EDIT ONLY THIS LINE)
# ---------------------------
IMAGE_PATH = r"C:\Users\Yioulika Antoniades\Desktop\UDA_Datasets\Dataset1_Classes\test\Animal Protein\Seafood\Fish\Fish_Test_3.jpg"  # <-- CHANGE THIS ONLY

# ---------------------------
# OPTIONAL (ONLY EDIT IF YOUR SAVE PATHS DIFFER)
# ---------------------------
# Model 1 files (hierarchical classifier)
MODEL1_WEIGHTS_PATH = r"outputs\best_hierarchical_model.pth"
MODEL1_STATS_PATH   = r"outputs\model1_norm_stats.json"  # contains {"mean":[...], "std":[...]}

# Model 2 files (fruit condition classifier)
MODEL2_WEIGHTS_PATH = r"outputs\best_fruit_condition_cnn.pt"
MODEL2_STATS_PATH   = r"outputs\model2_norm_stats.json"  # contains {"mean":[...], "std":[...]}

# ---------------------------
# SAFETY CHECKS (fail early with clear messages)
# ---------------------------
assert os.path.exists(IMAGE_PATH), f"Image not found: {IMAGE_PATH}"

assert os.path.exists(MODEL1_WEIGHTS_PATH), f"Model 1 weights not found: {MODEL1_WEIGHTS_PATH}"
assert os.path.exists(MODEL1_STATS_PATH),   f"Model 1 stats JSON not found: {MODEL1_STATS_PATH}"

assert os.path.exists(MODEL2_WEIGHTS_PATH), f"Model 2 weights not found: {MODEL2_WEIGHTS_PATH}"
assert os.path.exists(MODEL2_STATS_PATH),   f"Model 2 stats JSON not found: {MODEL2_STATS_PATH}"

# ---------------------------
# DEVICE SETUP
# ---------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# ---------------------------
# CONFIG (kept minimal for inference)
# ---------------------------
@dataclass
class InferenceConfig:
    img_size: int = 160  # MUST match training img_size for both models

cfg = InferenceConfig()

# ============================================================
# SECTION A — LABELS (must match training exactly)
# ============================================================
# Model 1 (hierarchical) leaf classes:
# IMPORTANT: these must match your training order exactly.
# In your Model 1 code: LEAF_GLOBAL = FRUITS + VEGETABLES + MEAT_POULTRY + SEAFOOD
LEVEL1 = ["Fresh Produce", "Animal Protein"]
LEVEL2_FRESH = ["Fruits", "Vegetables"]
LEVEL2_ANIMAL = ["Meat & Poultry", "Seafood"]

FRUITS_1 = ["Apples", "Avocados", "Bananas", "Grapes", "Mangos"]
VEGETABLES_1 = ["Carrots", "Cucumbers", "Lettuce", "Potatoes", "Tomatoes"]
MEAT_POULTRY_1 = ["Beef", "Chicken"]
SEAFOOD_1 = ["Fish"]

LEAF_GLOBAL = FRUITS_1 + VEGETABLES_1 + MEAT_POULTRY_1 + SEAFOOD_1

# Model 2 (condition classifier) labels:
CONDITIONS = ["Fresh", "Rotten", "Formalin-mixed"]
idx_to_cond = {i: c for i, c in enumerate(CONDITIONS)}

# Model 2 is ONLY meaningful for these fruit names (Model 1 leaf output is plural)
SUPPORTED_LEAFS_FOR_MODEL2 = {"Apples", "Bananas", "Grapes", "Mangos"}

# ============================================================
# SECTION B — MODEL DEFINITIONS (must match training architecture)
# ============================================================

class HierarchicalCNN(nn.Module):
    """
    Model 1 (Hierarchical CNN):
      - Shared conv backbone
      - Heads:
          Level 1: 2 classes (Fresh Produce vs Animal Protein)
          Level 2: 2 classes (Fruits vs Vegetables OR Meat&Poultry vs Seafood)
          Leaf heads: 4 heads depending on branch (Fruits/Vegetables/Meat/Seafood)
    """
    def __init__(self):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 32, 3, padding=1), nn.BatchNorm2d(32), nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, 3, padding=1), nn.BatchNorm2d(64), nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            nn.Conv2d(64, 128, 3, padding=1), nn.BatchNorm2d(128), nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            nn.Conv2d(128, 256, 3, padding=1), nn.BatchNorm2d(256), nn.ReLU(inplace=True),
            nn.MaxPool2d(2),
        )
        self.pool = nn.AdaptiveAvgPool2d((1, 1))
        self.embed = nn.Sequential(
            nn.Flatten(),
            nn.Linear(256, 256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
        )

        self.head1 = nn.Linear(256, 2)
        self.head2 = nn.Linear(256, 2)

        self.fruit_head = nn.Linear(256, len(FRUITS_1))
        self.veg_head = nn.Linear(256, len(VEGETABLES_1))
        self.meat_head = nn.Linear(256, len(MEAT_POULTRY_1))
        self.seafood_head = nn.Linear(256, len(SEAFOOD_1))

    def forward(self, x):
        x = self.features(x)
        x = self.pool(x)
        h = self.embed(x)
        return (
            self.head1(h), self.head2(h),
            self.fruit_head(h), self.veg_head(h),
            self.meat_head(h), self.seafood_head(h)
        )


class Net(nn.Module):
    """
    Model 2 (Condition CNN):
      - Predicts condition only: Fresh / Rotten / Formalin-mixed
      - Architecture copied from your Model 2 code.
    """
    def __init__(self, num_classes: int = 3):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 16, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            nn.Conv2d(16, 64, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            nn.Conv2d(64, 128, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),
        )
        self.gap = nn.AdaptiveAvgPool2d((4, 4))
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(128 * 4 * 4, 256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
            nn.Linear(256, num_classes),
        )

    def forward(self, x):
        x = self.features(x)
        x = self.gap(x)
        x = self.classifier(x)
        return x

# ============================================================
# SECTION C — HELPER: Decode Model 1 outputs into strings
# ============================================================

@torch.no_grad()
def predict_hierarchy_from_outputs(out1, out2, lf, lv, lm, ls):
    """
    Converts raw logits from Model 1 into:
      - predicted Level 1 index per sample
      - predicted Level 2 index per sample
      - predicted leaf string per sample (e.g., "Fish", "Apples", ...)
    """
    pred1 = out1.argmax(dim=1)  # 0=Fresh Produce, 1=Animal Protein
    pred2 = out2.argmax(dim=1)  # 0/1 branch-dependent

    pred_strings = []
    for i in range(out1.size(0)):
        p1, p2 = pred1[i].item(), pred2[i].item()

        if p1 == 0:
            lvl1_str = "Fresh Produce"
            if p2 == 0:
                lvl2_str = "Fruits"
                leaf_str = FRUITS_1[lf[i].argmax().item()]
            else:
                lvl2_str = "Vegetables"
                leaf_str = VEGETABLES_1[lv[i].argmax().item()]
        else:
            lvl1_str = "Animal Protein"
            if p2 == 0:
                lvl2_str = "Meat & Poultry"
                leaf_str = MEAT_POULTRY_1[lm[i].argmax().item()]
            else:
                lvl2_str = "Seafood"
                leaf_str = SEAFOOD_1[ls[i].argmax().item()]

        pred_strings.append((lvl1_str, lvl2_str, leaf_str))

    return pred_strings

# ============================================================
# SECTION D — LOAD NORMALIZATION STATS (mean/std) FROM JSON
# ============================================================

def load_norm_stats(json_path: str) -> Tuple[List[float], List[float]]:
    """
    Loads normalization stats from a JSON file:
      {"mean":[m1,m2,m3], "std":[s1,s2,s3]}
    """
    with open(json_path, "r", encoding="utf-8") as f:
        d = json.load(f)
    mean = d["mean"]
    std = d["std"]
    assert len(mean) == 3 and len(std) == 3, "Mean/std must be 3-channel (RGB)."
    return mean, std

mean1, std1 = load_norm_stats(MODEL1_STATS_PATH)
mean2, std2 = load_norm_stats(MODEL2_STATS_PATH)

# ============================================================
# SECTION E — BUILD INFERENCE TRANSFORMS (same as EVAL pipeline)
# ============================================================

model1_infer_tf = transforms.Compose([
    transforms.Resize(cfg.img_size + 16),
    transforms.CenterCrop(cfg.img_size),
    transforms.ToTensor(),
    transforms.Normalize(mean=mean1, std=std1),
])

model2_infer_tf = transforms.Compose([
    transforms.Resize(cfg.img_size + 16),
    transforms.CenterCrop(cfg.img_size),
    transforms.ToTensor(),
    transforms.Normalize(mean=mean2, std=std2),
])

# ============================================================
# SECTION F — LOAD BOTH MODELS (weights + eval mode)
# ============================================================

hier_model = HierarchicalCNN().to(device)
hier_model.load_state_dict(torch.load(MODEL1_WEIGHTS_PATH, map_location=device))
hier_model.eval()

cond_model = Net(num_classes=3).to(device)
cond_model.load_state_dict(torch.load(MODEL2_WEIGHTS_PATH, map_location=device))
cond_model.eval()

# ============================================================
# SECTION G — RUN CASCADED INFERENCE ON ONE IMAGE
# ============================================================

img = Image.open(IMAGE_PATH).convert("RGB")

# ---- Step 1: Model 1 leaf prediction ----
x1 = model1_infer_tf(img).unsqueeze(0).to(device)

with torch.no_grad():
    out1, out2, lf, lv, lm, ls = hier_model(x1)
    leaf_strings = predict_hierarchy_from_outputs(out1, out2, lf, lv, lm, ls)

pred_lvl1, pred_lvl2, pred_leaf_name = leaf_strings[0]

print("\nMODEL 1 RESULT")
print("----------------------------")
print(f"Predicted Level 1  : {pred_lvl1}")
print(f"Predicted Level 2  : {pred_lvl2}")
print(f"Predicted Leaf     : {pred_leaf_name}")

# ---- Step 2: Gate (run Model 2 only if supported fruit leaf) ----
if pred_leaf_name not in SUPPORTED_LEAFS_FOR_MODEL2:
    print("\nMODEL 2 SKIPPED")
    print("----------------------------")
    print(f"Reason: '{pred_leaf_name}' is not supported by Model 2 (condition classifier).")
    print("\nFINAL CASCADED OUTPUT")
    print("----------------------------")
    print(f"Leaf class  : {pred_leaf_name}")
    print(f"Condition   : N/A")
else:
    # ---- Step 3: Model 2 condition prediction ----
    x2 = model2_infer_tf(img).unsqueeze(0).to(device)

    with torch.no_grad():
        out = cond_model(x2)
        cond_idx = out.argmax(dim=1).item()
        cond_name = idx_to_cond[cond_idx]

    print("\nMODEL 2 RESULT")
    print("----------------------------")
    print(f"Predicted condition : {cond_name}")

    print("\nFINAL CASCADED OUTPUT")
    print("----------------------------")
    print(f"Leaf class  : {pred_leaf_name}")
    print(f"Condition   : {cond_name}")
```
